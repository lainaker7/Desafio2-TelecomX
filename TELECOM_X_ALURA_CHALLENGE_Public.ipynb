```python
# Análisis rápido de calidad de datos
def quick_data_quality_check(df):
    print("🔍 ANÁLISIS RÁPIDO DE CALIDAD DE DATOS")
    print("="*60)

    results = {}

    # 1. Valores nulos
    null_sum = df.isnull().sum().sum()
    results['null_values'] = null_sum
    print(f"✅ Valores nulos: {null_sum}")

    # 2. Duplicados
    duplicates = df.duplicated().sum()
    results['duplicates'] = duplicates
    print(f"✅ Duplicados completos: {duplicates}")

    # 3. Duplicados en ID
    if 'customerID' in df.columns:
        id_duplicates = df['customerID'].duplicated().sum()
        results['id_duplicates'] = id_duplicates
        print(f"✅ Duplicados en customerID: {id_duplicates}")

    # 4. Tipos de datos incorrectos
    if 'account.Charges.Total' in df.columns:
        if df['account.Charges.Total'].dtype == 'object':
            results['format_errors'] = "Total Charges es object, debería ser numérico"
            print(f"⚠️  account.Charges.Total: tipo incorrecto (object)")

    # 5. Rangos imposibles
    if 'customer.tenure' in df.columns:
        negative_tenure = (df['customer.tenure'] < 0).sum()
        results['negative_tenure'] = negative_tenure
        if negative_tenure > 0:
            print(f"⚠️  tenure negativo: {negative_tenure} registros")

    # 6. Valores categóricos inesperados
    unexpected_counts = {}
    expected_churn = ['Yes', 'No']
    if 'Churn' in df.columns:
        unexpected_churn = set(df['Churn'].unique()) - set(expected_churn)
        if unexpected_churn:
            results['unexpected_churn'] = unexpected_churn
            print(f"⚠️  Valores inesperados en Churn: {unexpected_churn}")

    print("="*60)
    return results

# Ejecutar análisis rápido
quality_report = quick_data_quality_check(df)
```

```python
def traducir_contenido_inteligente(df):
    """
    Traducción inteligente con lógica específica por columna
    """
    df_t = df.copy()

    # 1. Traducir Género
    if 'Género' in df_t.columns:
        df_t['Género'] = df_t['Género'].apply(
            lambda x: 'Hombre' if str(x).lower() in ['male', 'm', 'hombre']
            else 'Mujer' if str(x).lower() in ['female', 'f', 'mujer']
            else x
        )

    # 2. Traducir variables Sí/No
    columnas_si_no = ['Deserción', 'Socios', 'Dependendientes', 'Servicio_Teléfono',
                     'Líneas_Multiples', 'Facturación_sin_papel']

    for col in columnas_si_no:
        if col in df_t.columns:
            df_t[col] = df_t[col].apply(
                lambda x: 'Sí' if str(x).lower() in ['yes', 'si', 'sí', 'true', 'verdadero']
                else 'No' if str(x).lower() in ['no', 'no', 'false', 'falso']
                else x
            )

    # 3. Traducir Contrato
    if 'Contrato' in df_t.columns:
        mapeo_contrato = {
            'Month-to-month': 'Mensual',
            'One year': 'Anual',
            'Two year': 'Bianual',
            'month-to-month': 'Mensual',
            'one year': 'Anual',
            'two year': 'Bianual'
        }
        df_t['Contrato'] = df_t['Contrato'].replace(mapeo_contrato)

    # 4. Traducir Servicio de Internet
    if 'Servicio_Internet' in df_t.columns:
        mapeo_internet = {
            'Fiber optic': 'Fibra óptica',
            'fiber optic': 'Fibra óptica',
            'DSL': 'DSL',
            'dsl': 'DSL',
            'No': 'No tiene',
            'no': 'No tiene',
            'No internet service': 'Sin servicio'
        }
        df_t['Servicio_Internet'] = df_t['Servicio_Internet'].replace(mapeo_internet)

    # 5. Traducir Método de Pago
    if 'Metodo_de_pago' in df_t.columns:
        mapeo_pago = {
            'Mailed check': 'Cheque por correo',
            'Electronic check': 'Cheque electrónico',
            'Bank transfer (automatic)': 'Transferencia bancaria',
            'Credit card (automatic)': 'Tarjeta de crédito',
            'mailed check': 'Cheque por correo',
            'electronic check': 'Cheque electrónico'
        }
        df_t['Metodo_de_pago'] = df_t['Metodo_de_pago'].replace(mapeo_pago)

    # 6. Servicios adicionales (manejar "No internet service")
    servicios_internet = ['Seguridad_Online', 'Respaldo_online', 'Proteccion_dispositivo',
                         'Soporte_Tecnico', 'StreamingTV', 'StreamingMovies']

    for servicio in servicios_internet:
        if servicio in df_t.columns:
            # Primero manejar "No internet service"
            df_t[servicio] = df_t[servicio].replace(
                {'No internet service': 'No aplica', 'No Internet Service': 'No aplica'}
            )
            # Luego traducir Yes/No
            df_t[servicio] = df_t[servicio].replace({'Yes': 'Sí', 'No': 'No'})

    return df_t

# Ejecutar
df_completo_espanol = traducir_contenido_inteligente(df)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Configuración básica
plt.style.use('seaborn-v0_8-darkgrid')
pd.set_option('display.float_format', lambda x: '%.2f' % x)

def analisis_descriptivo_resumido(df_completo_espanol):
    """
    Análisis descriptivo resumido con métricas clave
    """
    print("=" * 70)
    print("📊 ANÁLISIS DESCRIPTIVO RESUMIDO")
    print("=" * 70)

    # 1. ESTADÍSTICAS BÁSICAS DEL DATASET
    print("\n1. 📋 INFORMACIÓN GENERAL:")
    print(f"   • Total de registros: {df_completo_espanol.shape[0]:,}")
    print(f"   • Total de variables: {df_completo_espanol.shape[1]}")
    print(f"   • Variables numéricas: {len(df_completo_espanol.select_dtypes(include=[np.number]).columns)}")
    print(f"   • Variables categóricas: {len(df_completo_espanol.select_dtypes(include=['object']).columns)}")

    # 2. ANÁLISIS DE VARIABLES NUMÉRICAS (MÉTRICAS CLAVE)
    variables_numericas = df_completo_espanol.select_dtypes(include=[np.number]).columns

    if len(variables_numericas) > 0:
        print("\n2. 🔢 ESTADÍSTICAS DE VARIABLES NUMÉRICAS:")
        print("-" * 70)

        # Crear DataFrame con métricas
        metricas = pd.DataFrame(index=variables_numericas)

        for col in variables_numericas:
            metricas.loc[col, 'Media'] = df_completo_espanol[col].mean()
            metricas.loc[col, 'Mediana'] = df_completo_espanol[col].median()
            metricas.loc[col, 'Desv. Estándar'] = df_completo_espanol[col].std()
            metricas.loc[col, 'Mínimo'] = df_completo_espanol[col].min()
            metricas.loc[col, 'Máximo'] = df_completo_espanol[col].max()
            metricas.loc[col, 'Q1 (25%)'] = df_completo_espanol[col].quantile(0.25)
            metricas.loc[col, 'Q3 (75%)'] = df_completo_espanol[col].quantile(0.75)
            metricas.loc[col, 'Rango'] = df_completo_espanol[col].max() - df_completo_espanol[col].min()
            metricas.loc[col, 'Coef. Variación (%)'] = (df_completo_espanol[col].std() / df_completo_espanol[col].mean() * 100) if df_completo_espanol[col].mean() != 0 else np.nan

        print(metricas.round(2))

        # 3. ANÁLISIS DE DISTRIBUCIÓN
        print("\n3. 📈 ANÁLISIS DE DISTRIBUCIÓN (Asimetría y Curtosis):")
        distribucion = pd.DataFrame(index=variables_numericas)
        distribucion['Asimetría'] = df_completo_espanol[variables_numericas].skew()
        distribucion['Curtosis'] = df_completo_espanol[variables_numericas].kurtosis()

        # Interpretación
        distribucion['Tipo Asimetría'] = distribucion['Asimetría'].apply(
            lambda x: 'Simétrica' if abs(x) < 0.5 else
                     'Moderada (+) derecha' if x >= 0.5 else
                     'Moderada (-) izquierda'
        )

        distribucion['Tipo Curtosis'] = distribucion['Curtosis'].apply(
            lambda x: 'Normal' if abs(x) < 0.5 else
                     'Leptocúrtica (picuda)' if x >= 0.5 else
                     'Platicúrtica (aplanada)'
        )

        print(distribucion[['Asimetría', 'Tipo Asimetría', 'Curtosis', 'Tipo Curtosis']].round(3))

    # 4. ANÁLISIS DE VARIABLES CATEGÓRICAS (RESUMIDO)
    variables_categoricas = df_completo_espanol.select_dtypes(include=['object']).columns

    if len(variables_categoricas) > 0:
        print("\n4. 📊 ESTADÍSTICAS DE VARIABLES CATEGÓRICAS:")
        print("-" * 70)

        resumen_categoricas = pd.DataFrame(index=variables_categoricas)

        for col in variables_categoricas:
            resumen_categoricas.loc[col, 'Valores Únicos'] = df_completo_espanol[col].nunique()
            moda = df_completo_espanol[col].mode()[0] if not df_completo_espanol[col].mode().empty else 'N/A'
            frecuencia_moda = (df_completo_espanol[col] == moda).sum()
            porcentaje_moda = (frecuencia_moda / len(df_completo_espanol)) * 100

            resumen_categoricas.loc[col, 'Moda'] = moda
            resumen_categoricas.loc[col, 'Frec. Moda'] = frecuencia_moda
            resumen_categoricas.loc[col, '% Moda'] = f"{porcentaje_moda:.1f}%"

        print(resumen_categoricas)

    # 5. ANÁLISIS ESPECÍFICO DE VARIABLES CLAVE
    print("\n5. 🎯 ANÁLISIS DE VARIABLES CLAVE:")
    print("-" * 70)

    # Variable 1: Antigüedad (si existe)
    if 'Antiguedad' in df.columns:
        print(f"\n📅 ANTIGÜEDAD (meses):")
        print(f"   • Media: {df_completo_espanol['Antiguedad'].mean():.1f} meses")
        print(f"   • Mediana: {df_completo_espanol['Antiguedad'].median():.1f} meses")
        print(f"   • Desv. Estándar: {df_completo_espanol['Antiguedad'].std():.1f} meses")
        print(f"   • Rango: {df_completo_espanol['Antiguedad'].min()} - {df_completo_espanol['Antiguedad'].max()} meses")

        # Segmentación por antigüedad
        segmentos = pd.cut(df_completo_espanol['Antiguedad'],
                          bins=[0, 12, 24, 36, df['Antiguedad'].max()],
                          labels=['0-1 año', '1-2 años', '2-3 años', '3+ años'])

        print(f"   • Distribución por segmentos:")
        for segmento in ['0-1 año', '1-2 años', '2-3 años', '3+ años']:
            cantidad = (segmentos == segmento).sum()
            porcentaje = (cantidad / len(df_completo_espanol)) * 100
            print(f"     - {segmento}: {cantidad} clientes ({porcentaje:.1f}%)")

    # Variable 2: Cargos Mensuales (si existe)
    if 'Cargos_mensuales' in df_completo_espanol.columns:
        print(f"\n💰 CARGOS MENSUALES:")
        print(f"   • Media: ${df_completo_espanol['Cargos_mensuales'].mean():.2f}")
        print(f"   • Mediana: ${df_completo_espanol['Cargos_mensuales'].median():.2f}")
        print(f"   • Desv. Estándar: ${df_completo_espanol['Cargos_mensuales'].std():.2f}")
        print(f"   • Rango: ${df_completo_espanol['Cargos_mensuales'].min():.2f} - ${df_completo_espanol['Cargos_mensuales'].max():.2f}")

        # Segmentos de valor
        cuartiles = df_completo_espanol['Cargos_mensuales'].quantile([0.25, 0.5, 0.75])
        print(f"   • Cuartiles: Q1=${cuartiles[0.25]:.2f}, Q2=${cuartiles[0.5]:.2f}, Q3=${cuartiles[0.75]:.2f}")

    # Variable 3: Deserción (si existe)
    if 'Deserción' in df_completo_espanol.columns:
        print(f"\n📉 DESERCIÓN:")
        distribucion_desercion = df_completo_espanol['Deserción'].value_counts()
        porcentajes = (df_completo_espanol['Deserción'].value_counts(normalize=True) * 100).round(1)

        for valor in distribucion_desercion.index:
            print(f"   • {valor}: {distribucion_desercion[valor]} ({porcentajes[valor]}%)")

    # 6. CORRELACIONES ENTRE VARIABLES NUMÉRICAS CLAVE
    if len(variables_numericas) >= 2:
        print("\n6. 🔗 CORRELACIONES ENTRE VARIABLES NUMÉRICAS:")
        print("-" * 70)

        # Seleccionar solo algunas variables clave para correlación
        vars_clave = [col for col in ['Antiguedad', 'Cargos_mensuales', 'Cuentas_Diarias']
                     if col in variables_numericas]

        if len(vars_clave) >= 2:
            corr_matrix = df[vars_clave].corr()
            print(corr_matrix.round(3))

            # Interpretación de correlaciones fuertes
            print("\n   Interpretación de correlaciones:")
            for i in range(len(corr_matrix.columns)):
                for j in range(i+1, len(corr_matrix.columns)):
                    corr = corr_matrix.iloc[i, j]
                    if abs(corr) > 0.5:
                        fuerza = "Fuerte" if abs(corr) > 0.7 else "Moderada"
                        signo = "positiva" if corr > 0 else "negativa"
                        print(f"   • {corr_matrix.columns[i]} y {corr_matrix.columns[j]}: "
                              f"{fuerza} correlación {signo} ({corr:.3f})")

    # 7. VISUALIZACIONES CLAVE
    print("\n7. 📊 VISUALIZACIONES RESUMIDAS:")
    print("-" * 70)

    fig, axes = plt.subplots(2, 2, figsize=(14, 10))

    # Gráfico 1: Distribución de antigüedad
    if 'Antiguedad' in df_completo_espanol.columns:
        ax1 = axes[0, 0]
        ax1.hist(df_completo_espanol['Antiguedad'], bins=30, edgecolor='black', alpha=0.7, color='skyblue')
        ax1.axvline(df_completo_espanol['Antiguedad'].mean(), color='red', linestyle='--', linewidth=2, label=f'Media: {df["Antiguedad"].mean():.1f}')
        ax1.axvline(df_completo_espanol['Antiguedad'].median(), color='green', linestyle='--', linewidth=2, label=f'Mediana: {df["Antiguedad"].median():.1f}')
        ax1.set_xlabel('Meses de Antigüedad')
        ax1.set_ylabel('Frecuencia')
        ax1.set_title('Distribución de Antigüedad', fontweight='bold')
        ax1.legend()
        ax1.grid(True, alpha=0.3)

    # Gráfico 2: Distribución de cargos mensuales
    if 'Cargos_mensuales' in df_completo_espanol.columns:
        ax2 = axes[0, 1]
        ax2.hist(df_completo_espanol['Cargos_mensuales'], bins=30, edgecolor='black', alpha=0.7, color='lightcoral')
        ax2.axvline(df_completo_espanol['Cargos_mensuales'].mean(), color='red', linestyle='--', linewidth=2, label=f'Media: ${df["Cargos_mensuales"].mean():.2f}')
        ax2.axvline(df_completo_espanol['Cargos_mensuales'].median(), color='green', linestyle='--', linewidth=2, label=f'Mediana: ${df["Cargos_mensuales"].median():.2f}')
        ax2.set_xlabel('Cargos Mensuales ($)')
        ax2.set_ylabel('Frecuencia')
        ax2.set_title('Distribución de Cargos Mensuales', fontweight='bold')
        ax2.legend()
        ax2.grid(True, alpha=0.3)

    # Gráfico 3: Boxplot de cargos por deserción
    if 'Deserción' in df_completo_espanol.columns and 'Cargos_mensuales' in df_completo_espanol.columns:
        ax3 = axes[1, 0]
        df.boxplot(column='Cargos_mensuales', by='Deserción', ax=ax3, grid=False)
        ax3.set_xlabel('Deserción')
        ax3.set_ylabel('Cargos Mensuales ($)')
        ax3.set_title('Cargos Mensuales vs Deserción', fontweight='bold')
        ax3.grid(True, alpha=0.3)

    # Gráfico 4: Distribución de deserción
    if 'Deserción' in df_completo_espanol.columns:
        ax4 = axes[1, 1]
        counts = df_completo_espanol['Deserción'].value_counts()
        colors = ['lightgreen', 'lightcoral'] if len(counts) == 2 else ['skyblue']
        ax4.bar(counts.index, counts.values, color=colors, edgecolor='black', alpha=0.7)

        # Agregar valores en las barras
        for i, v in enumerate(counts.values):
            ax4.text(i, v + max(counts.values)*0.01, str(v), ha='center', fontweight='bold')

        ax4.set_xlabel('Deserción')
        ax4.set_ylabel('Cantidad de Clientes')
        ax4.set_title('Distribución de Deserción', fontweight='bold')
        ax4.grid(True, alpha=0.3)

    plt.suptitle('ANÁLISIS DESCRIPTIVO RESUMIDO - MÉTRICAS CLAVE',
                fontsize=14, fontweight='bold', y=1.02)
    plt.tight_layout()
    plt.show()

    # 8. CONCLUSIONES PRINCIPALES
    print("\n8. 🎯 CONCLUSIONES PRINCIPALES:")
    print("-" * 70)

    conclusiones = []

    # Conclusión sobre deserción
    if 'Deserción' in df_completo_espanol.columns:
        tasa_desercion = df_completo_espanol['Deserción'].value_counts(normalize=True).get('Sí',
                     df_completo_espanol['Deserción'].value_counts(normalize=True).get('Yes', 0))
        if tasa_desercion > 0:
            conclusiones.append(f"• Tasa de deserción: {tasa_desercion*100:.1f}%")

    # Conclusión sobre antigüedad
    if 'Antiguedad' in df_completo_espanol.columns:
        conclusiones.append(f"• Antigüedad promedio: {df_completo_espanol['Antiguedad'].mean():.1f} meses")

        # Porcentaje de clientes nuevos (< 1 año)
        clientes_nuevos = (df_completo_espanol['Antiguedad'] < 12).sum()
        porcentaje_nuevos = (clientes_nuevos / len(df_completo_espanol)) * 100
        conclusiones.append(f"• Clientes con menos de 1 año: {porcentaje_nuevos:.1f}%")

    # Conclusión sobre cargos
    if 'Cargos_mensuales' in df_completo_espanol.columns:
        conclusiones.append(f"• Cargo mensual promedio: ${df_completo_espanol['Cargos_mensuales'].mean():.2f}")

        # Variabilidad
        coef_variacion = (df_completo_espanol['Cargos_mensuales'].std() / df_completo_espanol['Cargos_mensuales'].mean()) * 100
        conclusiones.append(f"• Variabilidad en cargos: {coef_variacion:.1f}% (CV)")

    # Conclusión sobre distribución
    if 'Antiguedad' in df_completo_espanol.columns:
        skewness = df_completo_espanol['Antiguedad'].skew()
        if skewness > 0.5:
            conclusiones.append("• La antigüedad está sesgada a la derecha (muchos clientes nuevos)")
        elif skewness < -0.5:
            conclusiones.append("• La antigüedad está sesgada a la izquierda (muchos clientes antiguos)")
        else:
            conclusiones.append("• La antigüedad tiene distribución aproximadamente simétrica")

    # Mostrar conclusiones
    for i, conclusion in enumerate(conclusiones, 1):
        print(f"  {i}. {conclusion}")

    print("\n" + "=" * 70)
    print("✅ ANÁLISIS DESCRIPTIVO COMPLETADO")
    print("=" * 70)

# EJECUTAR EL ANÁLISIS
analisis_descriptivo_resumido(df_completo_espanol)
```
