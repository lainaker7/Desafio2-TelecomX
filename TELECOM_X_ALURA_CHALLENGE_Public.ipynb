```python
# Si el JSON es directamente un array de objetos
url = "https://raw.githubusercontent.com/ingridcristh/challenge2-data-science-LATAM/refs/heads/main/TelecomX_Data.json"
df = pd.read_json(url)

# Mostrar las primeras filas
df.head()
```

```python
# Análisis rápido de calidad de datos
def quick_data_quality_check(df):
    print("🔍 ANÁLISIS RÁPIDO DE CALIDAD DE DATOS")
    print("="*60)

    results = {}

    # 1. Valores nulos
    null_sum = df.isnull().sum().sum()
    results['null_values'] = null_sum
    print(f"✅ Valores nulos: {null_sum}")

    # 2. Duplicados
    duplicates = df.duplicated().sum()
    results['duplicates'] = duplicates
    print(f"✅ Duplicados completos: {duplicates}")

    # 3. Duplicados en ID
    if 'customerID' in df.columns:
        id_duplicates = df['customerID'].duplicated().sum()
        results['id_duplicates'] = id_duplicates
        print(f"✅ Duplicados en customerID: {id_duplicates}")

    # 4. Tipos de datos incorrectos
    if 'account.Charges.Total' in df.columns:
        if df['account.Charges.Total'].dtype == 'object':
            results['format_errors'] = "Total Charges es object, debería ser numérico"
            print(f"⚠️  account.Charges.Total: tipo incorrecto (object)")

    # 5. Rangos imposibles
    if 'customer.tenure' in df.columns:
        negative_tenure = (df['customer.tenure'] < 0).sum()
        results['negative_tenure'] = negative_tenure
        if negative_tenure > 0:
            print(f"⚠️  tenure negativo: {negative_tenure} registros")

    # 6. Valores categóricos inesperados
    unexpected_counts = {}
    expected_churn = ['Yes', 'No']
    if 'Churn' in df.columns:
        unexpected_churn = set(df['Churn'].unique()) - set(expected_churn)
        if unexpected_churn:
            results['unexpected_churn'] = unexpected_churn
            print(f"⚠️  Valores inesperados en Churn: {unexpected_churn}")

    print("="*60)
    return results

# Ejecutar análisis rápido
quality_report = quick_data_quality_check(df)
```

## 📝 *Se revisan los nombres de cada columna y se eliminan y renombran para poder darles un mejor tratamiento*

```python
# Crear un diccionario de renombrado
rename_dict = {
    'customerID': 'ID_cliente',
    'Churn': 'Deserción',
    'customer.gender': 'Género',
    'customer.SeniorCitizen': 'Jubilado',
    'customer.Partner': 'Socios',
    'customer.Dependents': 'Dependendientes',
    'customer.tenure': 'Antiguedad',
    'phone.PhoneService': 'Servicio_Teléfono',
    'phone.MultipleLines': 'Líneas_Multiples',
    'internet.InternetService': 'Servicio_Internet',
    'internet.OnlineSecurity': 'Seguridad_Online',
    'internet.OnlineBackup': 'Respaldo_online',
    'internet.DeviceProtection': 'Proteccion_dispositivo',
    'internet.TechSupport': 'Soporte_Tecnico',
    'internet.StreamingTV': 'StreamingTV',
    'internet.StreamingMovies': 'StreamingMovies',
    'account.Contract': 'Contrato',
    'account.PaperlessBilling': 'Facturación_sin_papel',
    'account.PaymentMethod': 'Metodo_de_pago',
    'account.Charges.Monthly': 'Cargos_mensuales',
    'account.Charges.Total': 'Cargos_totales'
}
df = df.rename(columns=rename_dict)
df.info()
```

```python

df.columns.tolist() # Visualizamos los nombres de los encabezados en lista
```

```python

# Identificamos si hay valores vacios
# Encontrar cadenas vacías ('')
cadenas_vacias = (df == '').sum()
print("Cadenas vacías ('') por columna:")
print(cadenas_vacias[cadenas_vacias > 0])
```

```python
# Reemplazar '', ' ', '   ' → np.nan en TODAS las columnas string
df = df.replace(r'^\s*$', np.nan, regex=True)
```

```python
df['Cargos_totales'] = pd.to_numeric(df['Cargos_totales'], errors='coerce')
df.info()
```

```python
# Detectar columnas con valores NaN después del reemplazo
columnas_nulas = df.columns[df.isnull().any()].tolist()
print("Columnas con nulls:", columnas_nulas)
```

```python
for col in columnas_nulas:
    nulos = df[col].isnull().sum()
    porcentaje = df[col].isnull().mean() * 100

    print(f"Columna: {col}")
    print(f" - Cantidad de nulos: {nulos:,}")
    print(f" - Porcentaje: {porcentaje:.4f}%\n")
```

```python

# Eliminar registros vacios o nulls
df = df.dropna(subset=['Cargos_totales', 'Deserción'])
df.info()
```
